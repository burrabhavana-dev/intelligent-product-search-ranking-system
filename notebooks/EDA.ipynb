```python
df["search_text"] = (
    df["title"].fillna("") + " " +
    df["brand"].fillna("") + " " +
    df["category"].fillna("") + " " +
    df["specifications"].fillna("") + " " +
    df["description"].fillna("")
)
```

```python
important_columns = [
    "title",
    "brand",
    "category",
    "specifications",
    "description",
    "average_rating",
    "num_ratings",
    "discounted_price"
]

df[important_columns].head()
```

```python
df[important_columns].isnull().sum()
```

```python
text_columns = [
    "brand",
    "description",
    "specifications"
]

for col in text_columns:
    df[col] = df[col].fillna("")
```

```python
df["average_rating"] = df["average_rating"].fillna(0)

df["num_ratings"] = df["num_ratings"].fillna(0)
```

```python
df["search_text"] = (
    df["title"] + " " +
    df["brand"] + " " +
    df["category"] + " " +
    df["specifications"] + " " +
    df["description"]
)
```

```python
df["search_text"].head()
```

```python
import re

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r'[^a-z0-9\s]', ' ', text)

    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
df["clean_search_text"] = df["search_text"].apply(clean_text)
```

```python
df[
    ["title",
     "clean_search_text"]
].head()
```

```python
df.to_csv(
    "../data/clean_products.csv",
    index=False
)
```
